The following code does not perform addition in Scala or Python; it actually performs addition purely in Spark:

```python
df = spark.range(500).toDF("number")
df.select(df["number"] + 10)
```

Nothing has been computed yet: we only get back a new DataFrame with the column `(number + 10)`. An action such as `show()` makes Spark do the addition:

```python
df.select(df["number"] + 10).show(5)
```

```python
df.printSchema()
```

### Spark Types

To work with the correct Python types, use the following:

```python
from pyspark.sql.types import ByteType, StructType, StructField, StringType, LongType

b = ByteType()
```

```python
b
```

Types come together in a **schema**. Instead of letting Spark infer the types (as we did with `inferSchema` in the previous notebook), we can declare them explicitly:

```python
schema = StructType([
    StructField("country", StringType(), nullable=False),
    StructField("flights", LongType(), nullable=True),
])

countries = spark.createDataFrame([("United States", 411352), ("Canada", 8399)], schema)
countries.printSchema()
countries.show()
```
